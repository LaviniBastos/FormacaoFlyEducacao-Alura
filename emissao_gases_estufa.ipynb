# Agrupamento de dados

- Alterando formato do Dataframe

1. Uma única coluna com as informações dos anos
2. Apenas uma coluna com os dados de emissão


Para realizar as agregações nos dados de emissão, é necessário que a base de dados esteja em um formato adequado, com os valores de emissão todos em uma mesmo colunas. Para isso vamos modificar o formato da base de dados de forma que as colunas referentes ao ano se tornem uma só coluna.

```python
#Mantendo as colunas de 'Nível 1 setor', até a coluna 'Produto'

emissoes_gases.loc[:,'Nível 1 - Setor':'Produto'].columns
```

```python
colunas_info = list(emissoes_gases.loc[:,'Nível 1 - Setor':'Produto'].columns)
```

```python
emissoes_gases.loc[:,1970:2021].columns
```

```python
colunas_emissao = list(emissoes_gases.loc[:,1970:2021].columns)
colunas_emissao
```

```python
# Utilizando o método Melt para transformar

emissoes_gases.melt(id_vars=colunas_info, value_vars=colunas_emissao, var_name = 'Ano', value_name = 'Emissao')
```

```python
emissoes_por_ano = emissoes_gases.melt(id_vars=colunas_info, value_vars=colunas_emissao, var_name = 'Ano', value_name = 'Emissao')
```

```python
## Agora conseguiremos ter o numero total de gases para cada tipo de gás e identificar os mais emitidos

emissoes_por_ano.groupby('Gás')
```

```python
#Metodo sum() para saber a quantidade total de cada tipo de gás

emissoes_por_ano.groupby('Gás')[['Emissao']].sum()
```

```python
# Ordenando os dados com base em uma coluna

emissoes_por_gas = emissoes_por_ano.groupby('Gás')[['Emissao']].sum().sort_values('Emissao', ascending = False)
emissoes_por_gas
```

importando uma nova base de dados da população do censo do IBGE para fazer uma cruzamento entre as bases e descobrir a emissão per capita de cada estado.

```python
dados_ibge = pd.read_excel('/content/POP2022_Municipios.xls', header = 1, skipfooter=34)

dados_ibge
```

```python
### Como as informações são referentes aos municipios, vamos realizar uma agregação com base na coluna estados

dados_ibge.groupby('UF').sum(numeric_only = True)
```

```python
# A informação da populaçao não foi agregada, portanto será necessário algum tratamento pois ela não foi entendida como numérica

dados_ibge['POPULAÇÃO'].astype(int)
```

```python
## Existe caracteres nessa coluna que estão fazendo com que ela seja interpretada como string

dados_ibge[dados_ibge['POPULAÇÃO'].str.contains('\(', na = False)]
```

```python
## Conseguimos ver que existem mais campos com esse problema, então o tratamento precisa ser feito em todos eles com o método replace

# Co método assing permite criar outras colunas a partir de outras colunas.

dados_ibge = dados_ibge.assign(populacao_sem_parenteses = dados_ibge['POPULAÇÃO'].replace('\(\d{1,2}\)', '', regex = True),
                                populacao = lambda x : x.loc[:,'populacao_sem_parenteses'].replace('\.', '', regex = True)
                               )
```

```python
# Verificando se a transformação foi feita corretamente

dados_ibge[dados_ibge['POPULAÇÃO'].str.contains('\(', na = False)]
```

```python
#Agora podemos transforma a ultima coluna para ium tipo inteiro.

dados_ibge = dados_ibge.astype({'populacao': 'int64'})
```

```python
## agora conseguimos fazer o agrupamento

populacao_estados = dados_ibge.groupby('UF')[['populacao']].sum().reset_index()
populacao_estados
```

## Unindo os dados

```python
## Agora precisamos encontrar a seguinte informação: A Emissão total para cada um dos estados do ano mais recente
## Com as informações tratadas, podemos realizar a junção dos dados de emissão com a população, com o uso do método merge()

emissoes_estados = emissoes_por_ano[emissoes_por_ano['Ano'] == 2021].groupby('Estado')[['Emissao']].sum().reset_index()
emissoes_estados
```

```python
# Unindo as tabelas pela coluna UF que é a coluna em comum entre as duas

dados_agrupados = pd.merge(emissoes_estados, populacao_estados, left_on = 'Estado', right_on = 'UF')
dados_agrupados
```

```python
# Para verificar se existe uma relação entre o crescimento da população e o aumento da emissao de gases podemos utilizar o grafico de dispersão

dados_agrupados.plot(x = 'populacao', y = 'Emissao', kind = 'scatter', figsize=(8,6))
```

```python
# Será que existe uma forma de melhorar essa visualização, porque nela os valores estão em notação cientifica e não sabemos também quais os estados exatamente estão envolvidos.

import plotly.express as px

px.scatter(data_frame = dados_agrupados, x = 'populacao', y = 'Emissao', text= 'Estado', opacity = 0 )
```

```python
# Agora verificando a emissão per capita para cada um dos estados, criando uma nova coluna com a divisão dos valores de emissao pela populacao de cada estado.

dados_agrupados = dados_agrupados.assign(emissao_per_capita = dados_agrupados['Emissao']/dados_agrupados['populacao']).sort_values('emissao_per_capita', ascending = False)
dados_agrupados
```

```python
## Agora podemos criar uma visualização com a lib plotly

px.bar(data_frame = dados_agrupados, x = 'Estado', y = 'emissao_per_capita')
```

```python
## Agora melhorando mais ainda o gráfico de scatter feito anteriormente, essa modificação permitirá verificar tanto a populção, a emissão e a emissão per capita dos estados

px.scatter(data_frame = dados_agrupados, x = 'populacao', y = 'Emissao', text= 'Estado', size = 'emissao_per_capita' )
```

Conclusão: Com esse gráfico conseguimos ver que apesar do Pará ter uma emissão maior de gases mas com base na população a bolha dele não é tão grande, quando calculamos per capita quem está na frente e possui a bolha maior é o estado de Roraima
